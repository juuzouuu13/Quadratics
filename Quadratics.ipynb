```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
import ipywidgets as widgets


def basketball_simulation():

    x = np.linspace(2, 8, 100)
    y = -0.18 * (x - 5)**2 + 4.7

    fig, (court, graph) = plt.subplots(1, 2, figsize=(12, 5))

    fig.patch.set_facecolor("#e8c58b")
    court.set_facecolor("#e8c58b")
    graph.set_facecolor("white")


    court.plot([0, 10], [0, 0], color="#654321", linewidth=4)
    court.plot([8.5, 8.5], [2, 3.5], color="#555555", linewidth=6)
    court.plot([8.1, 8.5], [3.5, 3.5], color="white", linewidth=6)
    court.plot([8.1, 7.7], [3.2, 3.2], color="red", linewidth=4)
    court.plot([7.8, 7.95], [3.2, 2.7], color="white", linewidth=2)
    court.plot([8.0, 7.95], [3.2, 2.7], color="white", linewidth=2)

    court.plot(x, y, color="blue", linestyle="--", linewidth=2)

    court_ball = plt.Circle((x[0], y[0]), 0.25, color="#f28c28")
    court.add_patch(court_ball)

    court.set_xlim(0, 10)
    court.set_ylim(0, 5)
    court.set_aspect("equal")
    court.axis("off")


    graph.plot(x, y, color="blue", linewidth=2)
    graph_ball, = graph.plot([], [], "o", color="#f28c28", markersize=9)

    graph.set_xlim(0, 10)
    graph.set_ylim(0, 5)
    graph.set_xlabel("Horizontal distance")
    graph.set_ylabel("Height")
    graph.set_title("Basketball trajectory")
    graph.grid(alpha=0.3)


    def animate(i):
        court_ball.center = (x[i], y[i])
        graph_ball.set_data([x[i]], [y[i]])
        return court_ball, graph_ball

    ani = FuncAnimation(
        fig, animate,
        frames=len(x),
        interval=30,
        blit=True,
        repeat=True
    )

    animation_html = ani.to_jshtml()
    plt.close(fig)

    return HTML(animation_html)




def falling_ball_simulation():

    bounce_x = []
    bounce_y = []
    bounce_t = []
    bounce_number = []
    bounce_equations = []

    start_x = 0.8
    start_t = 0


    for i in range(6):

        duration = 1.4 * (0.72 ** i)
        distance = 2.0 * (0.72 ** i)
        peak = 2.8 * (0.65 ** i)

        u = np.linspace(0, 1, 30)

        arc_x = start_x + distance * u
        arc_y = 4 * peak * u * (1 - u)
        arc_t = start_t + duration * u

        if i > 0:
            arc_x = arc_x[1:]
            arc_y = arc_y[1:]
            arc_t = arc_t[1:]

        bounce_x.extend(arc_x)
        bounce_y.extend(arc_y)
        bounce_t.extend(arc_t)
        bounce_number.extend([i + 1] * len(arc_x))

        a = 4 * peak / duration**2
        midpoint = start_t + duration / 2

        equation = f"h = -{a:.2f}(t - {midpoint:.2f})² + {peak:.2f}"
        bounce_equations.extend([equation] * len(arc_x))

        start_x += distance
        start_t += duration

    bounce_x = np.array(bounce_x)
    bounce_y = np.array(bounce_y)
    bounce_t = np.array(bounce_t)
    bounce_number = np.array(bounce_number)
    bounce_equations = np.array(bounce_equations)


    fig, (scene, graph) = plt.subplots(1, 2, figsize=(12, 5))

    fig.patch.set_facecolor("#e8c58b")
    scene.set_facecolor("#e8c58b")

    scene.plot([0, 10], [0, 0], color="#654321", linewidth=4)

    scene.plot(
        bounce_x, bounce_y,
        color="blue", linewidth=2, linestyle="--"
    )

    ball = plt.Circle(
        (bounce_x[0], bounce_y[0]),
        0.22, color="#f28c28"
    )
    scene.add_patch(ball)

    scene.set_xlim(0, 10)
    scene.set_ylim(0, 4)
    scene.set_aspect("equal")
    scene.axis("off")

    bounce_label = scene.text(
        0.5, 3.5, "Bounce 1",
        fontsize=14, weight="bold"
    )

    height_label = scene.text(
        0.5, 3.15, "Height: 0.00",
        fontsize=12
    )


    graph.plot(bounce_t, bounce_y, color="blue", linewidth=2)

    graph_ball, = graph.plot(
        [], [], "o", color="#f28c28", markersize=9
    )

    graph.set_title("Height over time")
    graph.set_xlabel("Time")
    graph.set_ylabel("Height")
    graph.set_xlim(0, bounce_t[-1])
    graph.set_ylim(0, 3.5)
    graph.grid(alpha=0.3)

    equation_label = graph.text(
        0.04, 0.94, "",
        transform=graph.transAxes,
        fontsize=9,
        verticalalignment="top"
    )


    def animate(i):

        ball.center = (bounce_x[i], bounce_y[i])
        graph_ball.set_data([bounce_t[i]], [bounce_y[i]])

        bounce_label.set_text(f"Bounce {bounce_number[i]}")
        height_label.set_text(f"Height: {bounce_y[i]:.2f}")
        equation_label.set_text(bounce_equations[i])

        return (
            ball, graph_ball, bounce_label,
            height_label, equation_label
        )

    ani = FuncAnimation(
        fig,
        animate,
        frames=len(bounce_x),
        interval=30,
        blit=True,
        repeat=True
    )

    animation_html = ani.to_jshtml()
    plt.close(fig)

    return HTML(animation_html)






from IPython.display import HTML, display
import ipywidgets as widgets


basketball_html = basketball_simulation()
falling_html = falling_ball_simulation()


simulation = widgets.ToggleButtons(
    options=["Basketball", "Falling Ball"],
    description="Simulation:",
    button_style="info"
)

output = widgets.Output()

def switch_simulation(change):

    output.clear_output(wait=True)

    if change["new"] == "Basketball":
        selected = basketball_html

    else:
        selected = falling_html

    with output:
        display(selected)

simulation.observe(switch_simulation, names="value")

display(widgets.VBox([simulation, output]))


with output:
    display(basketball_html)
```
